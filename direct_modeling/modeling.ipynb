```python
import numpy as np
import matplotlib.pyplot as plt

class Curva:
    def __init__(self, x, y, label="Curva", color="blue"):
        self.x = np.array(x)
        self.y = np.array(y)
        self.label = label
        self.color = color

    """
    Si el usuario no pasa nada (curva.graficar()), entonces el método debe crear un nuevo gráfico (con plt.subplots()). 
    Si el usuario ya tiene un gráfico (fig, ax = plt.subplots()) y pasa ax (curva.graficar(ax)), entonces la curva se dibuja en ese gráfico."""
    
    def graficar(self, ax=None): #ax=None 
        if ax is None:
            fig, ax = plt.subplots()
        ax.plot(self.x, self.y, label=self.label, color=self.color)
        ax.legend()
        ax.axhline(0, color="black", linewidth=0.8)  # eje X
        #ax.axvline(0, color="black", linewidth=0.8)  # eje Y

        return ax
    
    
    def maximo(self):
        idx = np.argmax(self.y)
        return self.x[idx], self.y[idx]
    
    def minimo(self):
        idx = np.argmin(self.y)
        return self.x[idx], self.y[idx]

```
